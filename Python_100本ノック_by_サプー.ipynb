```python
class Student:
    def __init__(self, name:str , grade:int , section:str, scores:dict[str:int]):
        self.name = name
        self.grade = grade
        self.section = section
        self.scores = scores

    def add_score(self, subject, score):
        self.scores[subject] = score

    def total_score(self):
        return sum(self.scores.values())
```

```python
student = Student("鈴木", 2, "B")
print(student.name)

student.add_score("数学", 80)
student.add_score("英語", 70)
student.add_score("国語", 90)
print(student.scores)

print(student.total_score())
```

```python
class Card:
    def __init__(self, suit:str, number:int):
        self.suit = suit
        self.number = number

class Deck:
    def __init__(self):
        self.cards = []
        for suit in ["ハート", "ダイヤ", "スペード", "クラブ"]:
            for number in range(1, 14):
                card = Card(suit, number)
                self.cards.append(card)

    def shuffle(self):
        import random
        random.shuffle(self.cards)

    def draw_card(self):
        return self.cards.pop(0)

deck1 = Deck()
deck1.shuffle()
for _ in range(5):
    print(deck1.draw_card().suit,deck1.draw_card().number)
```
